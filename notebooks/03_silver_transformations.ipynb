```python
from pyspark.sql.functions import col, trim, lower

# Read Bronze table
bronze_deliveries = spark.table(
    "workspace.default.bronze_deliveries"
)

# Transform Bronze → Silver Candidate
silver_deliveries_candidate = bronze_deliveries.select(
    
    # Record identifiers
    trim(col("source_record_id")).alias("source_record_id"),
    trim(col("delivery_id")).alias("delivery_id"),
    trim(col("match_id")).alias("match_id"),

    # Delivery information
    col("innings_no").cast("int").alias("innings_no"),
    col("over_no").cast("int").alias("over_no"),
    col("ball_seq").cast("int").alias("ball_seq"),

    # Team / player identifiers
    trim(col("batting_team_id")).alias("batting_team_id"),
    trim(col("bowling_team_id")).alias("bowling_team_id"),
    trim(col("batter_id")).alias("batter_id"),
    trim(col("non_striker_id")).alias("non_striker_id"),
    trim(col("bowler_id")).alias("bowler_id"),

    # Runs
    col("runs_batter").cast("int").alias("runs_batter"),
    col("runs_extras").cast("int").alias("runs_extras"),
    col("total_runs").cast("int").alias("total_runs"),

    # Standardized categorical fields
    lower(trim(col("extras_type"))).alias("extras_type"),
    col("wicket_flag").cast("boolean").alias("wicket_flag"),
    lower(trim(col("dismissal_type"))).alias("dismissal_type"),

    # Dismissal / fielding information
    trim(col("dismissed_player_id")).alias("dismissed_player_id"),
    trim(col("fielder_id")).alias("fielder_id"),

    # Event timestamp
    col("event_ts").cast("timestamp").alias("event_ts"),

    # Lineage / ingestion metadata
    trim(col("source_file")).alias("source_file"),
    col("ingestion_timestamp").cast("timestamp").alias("ingestion_timestamp"),
    trim(col("ingestion_run_id")).alias("ingestion_run_id")
)

display(silver_deliveries_candidate.limit(10))
```

```python
from pyspark.sql.functions import col, trim, lower

# Read Bronze Matches
bronze_matches = spark.table(
    "workspace.default.bronze_matches"
)

# Bronze → Silver transformation
silver_matches_candidate = bronze_matches.select(

    # Identifiers
    trim(col("source_record_id")).alias("source_record_id"),
    trim(col("match_id")).alias("match_id"),

    # Match information
    col("season").cast("int").alias("season"),
    col("match_date").cast("date").alias("match_date"),
    col("scheduled_start_ts").cast("timestamp").alias("scheduled_start_ts"),

    # Venue
    trim(col("venue_id")).alias("venue_id"),

    # Teams
    trim(col("team_1_id").cast("string")).alias("team_1_id"),
    trim(col("team_1_name")).alias("team_1_name"),
    trim(col("team_2_id").cast("string")).alias("team_2_id"),
    trim(col("team_2_name")).alias("team_2_name"),

    # Toss
    trim(col("toss_winner_id").cast("string")).alias("toss_winner_id"),
    lower(trim(col("toss_decision"))).alias("toss_decision"),

    # Match result
    lower(trim(col("match_status"))).alias("match_status"),
    trim(col("winner_id").cast("string")).alias("winner_id"),
    lower(trim(col("result_type"))).alias("result_type"),

    # Match statistics
    col("winner_margin").cast("double").alias("winner_margin"),
    col("target_runs").cast("double").alias("target_runs"),
    col("innings_1_runs").cast("double").alias("innings_1_runs"),
    col("innings_1_wickets").cast("double").alias("innings_1_wickets"),
    col("innings_2_runs").cast("double").alias("innings_2_runs"),
    col("innings_2_wickets").cast("double").alias("innings_2_wickets"),

    # Player of the match
    trim(col("player_of_match_id")).alias("player_of_match_id"),

    # Lineage / ingestion metadata
    trim(col("source_file")).alias("source_file"),
    col("ingestion_timestamp").cast("timestamp").alias("ingestion_timestamp"),
    trim(col("ingestion_run_id")).alias("ingestion_run_id")
)

display(silver_matches_candidate.limit(10))
```

```python
from pyspark.sql.functions import col, trim, lower
from pyspark.sql.types import StringType

bronze_players = spark.table("workspace.default.bronze_players")

# Automatically clean string columns and preserve other types
expressions = []

for field in bronze_players.schema.fields:
    c = col(field.name)

    if isinstance(field.dataType, StringType):
        expressions.append(trim(c).alias(field.name))
    else:
        expressions.append(c.alias(field.name))

silver_players_candidate = bronze_players.select(*expressions)

display(silver_players_candidate.limit(10))
```

```python
print("Bronze count:", bronze_players.count())
print("Silver Candidate count:", silver_players_candidate.count())
print(
    "Counts match:",
    bronze_players.count() == silver_players_candidate.count()
)
```
